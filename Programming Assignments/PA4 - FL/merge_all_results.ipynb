```python
"""
Script to load and compare all experimental results from Tasks 2-4
Generates comprehensive plots and summary tables
"""

import pickle
import matplotlib.pyplot as plt
import numpy as np
from results_utils import load_results, plot_comparison, print_summary
```

```python
print("="*80)
print("LOADING ALL EXPERIMENTAL RESULTS")
print("="*80)

# Load results from each task
try:
    task2_varying_K = load_results('task2_varying_K')
    task2_sampling = load_results('task2_varying_sampling')
    task3_heterogeneity = load_results('task3_heterogeneity')
    task4_1_fedprox = load_results('task4_1_fedprox')
    task4_2_scaffold = load_results('task4_2_scaffold')  # Uncomment when implemented
    task4_3_fedgh = load_results('task4_3_fedgh')  # Uncomment when implemented
    task4_4_fedsam = load_results('task4_4_fedsam')  # Uncomment when implemented
    
    print("✓ All result files loaded successfully\n")
except FileNotFoundError as e:
    print(f"Error: {e}")
    print("Make sure you have run all experiments and saved results.")
    exit(1)

#%% Task 2 Summary
print("\n" + "="*80)
print("TASK 2: FEDAVG ANALYSIS")
print("="*80)
print_summary(task2_varying_K)
print_summary(task2_sampling)

#%% Task 3 Summary
print("\n" + "="*80)
print("TASK 3: HETEROGENEITY IMPACT")
print("="*80)
print_summary(task3_heterogeneity)

#%% Task 4 Comprehensive Comparison
print("\n" + "="*80)
print("TASK 4: ADVANCED METHODS COMPARISON")
print("="*80)

# Combine all Task 4 methods for comparison
# Use alpha=0.1 (high heterogeneity) for fair comparison
comparison_methods = {}

# Get FedAvg baseline (mu=0.0 from FedProx experiments)
if 'FedAvg' in task4_1_fedprox:
    comparison_methods['FedAvg'] = task4_1_fedprox['FedAvg']

# Get best FedProx
if 'FedProx (μ=0.01)' in task4_1_fedprox:
    comparison_methods['FedProx'] = task4_1_fedprox['FedProx (μ=0.01)']

# TODO: Add other methods when implemented
comparison_methods['SCAFFOLD'] = task4_2_scaffold['SCAFFOLD']
comparison_methods['FedGH'] = task4_3_fedgh['FedGH']
comparison_methods['FedSAM'] = task4_4_fedsam['FedSAM (ρ=0.05)']

# Plot comprehensive comparison
if comparison_methods:
    plot_comparison(comparison_methods, metric='test_accuracy',
                   title='Comparison of FL Methods (α=0.1, High Heterogeneity)',
                   save_path='./results/comprehensive_comparison.png')
    
    print_summary(comparison_methods)
```

```python
#%% Generate Final Summary Table
print("\n" + "="*80)
print("FINAL SUMMARY TABLE")
print("="*80)

summary_data = []

for method_name, results in comparison_methods.items():
    history = results['history'] if 'history' in results else results
    summary_data.append({
        'Method': method_name,
        'Final Accuracy (%)': f"{history['test_accuracy'][-1]:.2f}",
        'Best Accuracy (%)': f"{max(history['test_accuracy']):.2f}",
        'Final Loss': f"{history['test_loss'][-1]:.4f}"
    })

# Print as formatted table
print(f"\n{'Method':<20} {'Final Acc (%)':<15} {'Best Acc (%)':<15} {'Final Loss':<15}")
print("-" * 65)
for row in summary_data:
    print(f"{row['Method']:<20} {row['Final Accuracy (%)']:<15} {row['Best Accuracy (%)']:<15} {row['Final Loss']:<15}")

print("\n" + "="*80)
print("ALL RESULTS COMPILED!")
print("Check ./results/ directory for all plots")
print("="*80)
```
